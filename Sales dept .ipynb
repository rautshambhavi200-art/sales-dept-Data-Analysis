#### Univariate Analysis


```python
# Select only the actual numerical columns (excluding categorical/text columns like Date, Customer, City, etc.)
numeric_cols = ['Quantity', 'Unit_Price', 'Discount', 'Sales']

# Create subplots (2 rows x 2 columns for 4 variables)
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

# Loop through columns and plot
for i, col in enumerate(numeric_cols):
    sns.histplot(df[col], kde=True, ax=axes[i])

plt.tight_layout()
plt.show()
```

#### Bivariate and Multivariate Analysis

```python
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.show()
```
